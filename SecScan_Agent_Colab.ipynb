### 1. Descripción del Proyecto y Problemática
En el desarrollo de software moderno, los errores humanos introducen riesgos críticos de ciberseguridad:
- **Secretos expuestos:** Credenciales de nube (AWS, GCP), API keys y contraseñas subidas por descuido al repositorio.
- **Dependencias vulnerables:** Paquetes externos con CVEs activos.
- **Inyecciones SQL / XSS:** Código con concatenación insegura o renderizado no sanitizado.
- **Ataques activos en logs:** Intentos de fuerza bruta, escaneos de herramientas ofensivas (sqlmap, nikto) y path traversal.

**SecScan Agent** resuelve esto aplicando una arquitectura de Agente Inteligente basada en el ciclo **Observar ➔ Razonar ➔ Actuar ➔ Reportar**.

### 5. Visualización del Reporte HTML Generado
Renderizamos el informe interactivo con el resumen de severidades (Crítica, Alta, Media, Baja).

```python
import glob, os
import IPython.display as display

report_files = glob.glob('reports/security_report_*.html')
if report_files:
    latest = max(report_files, key=os.path.getctime)
    print(f'Mostrando: {latest}')
    with open(latest, 'r', encoding='utf-8') as f:
        html = f.read()
    display.display(display.HTML(html))
else:
    print('No se encontró archivo de reporte.')
```
